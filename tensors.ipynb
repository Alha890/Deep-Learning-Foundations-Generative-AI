```python
import torch
tensor = torch.randn(5, 4)
print("Original Tensor:\n", tensor)
print("Shape:", tensor.shape)
print("\nMean:", tensor.mean())
print("Max:", tensor.max())
print("Standard Deviation:", tensor.std())
reshaped_tensor = tensor.reshape(4, 5)
print("\nReshaped Tensor:\n", reshaped_tensor)
print("New Shape:", reshaped_tensor.shape)
```
